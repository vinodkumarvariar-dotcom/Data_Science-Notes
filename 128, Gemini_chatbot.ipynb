```python
import google.generativeai as genai

genai.configure(api_key="your GEMINI API KEY")

# ✅ Safe model
model = genai.GenerativeModel("gemini-2.5-flash")

def medical_chatbot():
    print("🩺 Medical Chatbot (type 'exit' to stop)\n")

    while True:
        user_input = input("You: ")

        if user_input.lower() == "exit":
            print("Chat ended.")
            break

        # 🚨 Emergency detection
        if any(word in user_input.lower() for word in ["chest pain", "breathing", "unconscious"]):
            print("🚨 Emergency! Please seek immediate medical help.")
            continue

        prompt = f"""
        You are a helpful medical assistant.
        Provide general health advice only.
        Do NOT give prescriptions.
        Suggest seeing a doctor when necessary.

        User: {user_input}
        """

        response = model.generate_content(prompt)
        print("Bot:", response.text)
        print("-" * 50)

medical_chatbot()
```
