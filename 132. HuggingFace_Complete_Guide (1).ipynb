### Core Architecture — Transformer Models
```
Input Text → Tokenizer → Token IDs → Model (Encoder/Decoder) → Output
```
- **Encoder-only** (e.g., BERT): Classification, NER, Q&A
- **Decoder-only** (e.g., GPT-2): Text generation
- **Encoder-Decoder** (e.g., T5, BART): Summarization, Translation

## ✍️ Section 3 — Text Generation

**Model used:** `gpt2`  
**Architecture:** Decoder-only (GPT-2)  
**Task:** Auto-regressive text completion — predict the next token repeatedly.

````python
print("📥 Loading text generation model: gpt2 ...")
generator = pipeline(
    "text-generation",
    model="gpt2",
    device=device
)
print("✅ Text generation model loaded!")
````

````python
prompts = [
    "In the future, artificial intelligence will",
    "The secret to learning machine learning is",
    "Once upon a time, a robot discovered that it could",
]

for prompt in prompts:
    output = generator(
        prompt,
        max_new_tokens=60,
        num_return_sequences=1,
        temperature=0.85,
        do_sample=True,
        pad_token_id=50256  # EOS token for GPT-2
    )
    print(f"\n📌 Prompt: '{prompt}'")
    print("─" * 60)
    generated = output[0]['generated_text']
    print(textwrap.fill(generated, width=80))
    print()
````

````python
# ── Generate multiple diverse completions ───────────────────────────────────
prompt = "Deep learning is revolutionizing"

outputs = generator(
    prompt,
    max_new_tokens=50,
    num_return_sequences=3,
    temperature=0.9,
    do_sample=True,
    pad_token_id=50256
)

print(f"✨ 3 different continuations for: '{prompt}'\n")
for i, out in enumerate(outputs, 1):
    print(f"  [{i}] {out['generated_text']}")
    print()
````

## 💬 Section 4 — Sentiment Analysis

**Model used:** `distilbert-base-uncased-finetuned-sst-2-english`  
**Architecture:** Encoder-only (DistilBERT)  
**Task:** Classify text as POSITIVE or NEGATIVE with a confidence score.

````python
print("📥 Loading sentiment analysis model ...")
sentiment = pipeline(
    "sentiment-analysis",
    model="distilbert-base-uncased-finetuned-sst-2-english",
    device=device
)
print("✅ Sentiment model loaded!")
````

````python
texts = [
    "I absolutely love learning about machine learning — it's fascinating!",
    "This model is incredibly slow and keeps giving wrong answers.",
    "The performance is okay, not great but not terrible either.",
    "Hugging Face has made NLP so much more accessible to everyone.",
    "I'm really disappointed with how this turned out.",
    "The transformer architecture changed everything in AI research."
]

results = sentiment(texts)

print("📊 Sentiment Analysis Results:\n")
print(f"{'Text':<55} {'Label':<12} {'Score'}")
print("─" * 80)

for text, result in zip(texts, results):
    label = result['label']
    score = result['score']
    emoji = "🟢" if label == "POSITIVE" else "🔴"
    short_text = text[:52] + "..." if len(text) > 52 else text
    bar = "█" * int(score * 20)
    print(f"{short_text:<55} {emoji} {label:<10} {score:.4f}  |{bar}|")
````

## ❓ Section 5 — Question Answering

**Model used:** `deepset/roberta-base-squad2`  
**Architecture:** Encoder-only (RoBERTa)  
**Task:** Extract the answer to a question from a given context passage.

````python
print("📥 Loading question answering model ...")
qa = pipeline(
    "question-answering",
    model="deepset/roberta-base-squad2",
    device=device
)
print("✅ QA model loaded!")
````

````python
context = """
The Transformer architecture was introduced in the landmark 2017 paper "Attention Is All You
Need" by researchers at Google Brain. Unlike previous sequence models that relied on
recurrent neural networks (RNNs) or convolutional networks, Transformers use a mechanism
called self-attention to process input sequences in parallel. This made training dramatically
faster and enabled models to capture long-range dependencies more effectively. BERT
(Bidirectional Encoder Representations from Transformers), released by Google in 2018,
demonstrated that pre-training a large Transformer on vast amounts of text followed by
fine-tuning on specific tasks could achieve state-of-the-art results across many NLP
benchmarks. GPT (Generative Pre-trained Transformer), developed by OpenAI, took the
decoder-only approach and achieved remarkable text generation capabilities.
"""

questions = [
    "When was the Transformer architecture introduced?",
    "What mechanism do Transformers use instead of RNNs?",
    "Who developed BERT?",
    "What does BERT stand for?",
    "Who developed GPT?"
]

print("📖 Context provided. Now answering questions...\n")

for q in questions:
    result = qa(question=q, context=context)
    print(f"❓ Q: {q}")
    print(f"✅ A: {result['answer']}  (confidence: {result['score']:.2%})")
    print()
````

## 🏷️ Section 7 — Zero-Shot Classification

**Model used:** `facebook/bart-large-mnli`  
**Task:** Classify text into *any* labels you define — **no fine-tuning needed!**  
This uses Natural Language Inference (NLI) under the hood.

````python
print("📥 Loading zero-shot classifier ...")
classifier = pipeline(
    "zero-shot-classification",
    model="facebook/bart-large-mnli",
    device=device
)
print("✅ Zero-shot classifier loaded!")
````

````python
# ── Example 1: News topic classification ─────────────────────────────────────
text = "The stock market crashed today amid concerns about rising inflation and interest rates."
candidate_labels = ["finance", "sports", "technology", "politics", "healthcare"]

result = classifier(text, candidate_labels)

print("📰 Zero-Shot News Classification:\n")
print(f"Text: \"{text}\"\n")
print("Scores:")
for label, score in zip(result['labels'], result['scores']):
    bar = "█" * int(score * 30)
    print(f"  {label:<15} {score:.4f}  |{bar}|")
````

````python
# ── Example 2: Multi-label classification ────────────────────────────────────
text2 = "The new Python library makes it easy to train neural networks on GPU clusters."
labels2 = ["programming", "machine learning", "cloud computing", "cooking", "travel"]

result2 = classifier(text2, labels2, multi_label=True)

print("\n🏷️ Multi-label Classification (multiple labels can be true):\n")
print(f"Text: \"{text2}\"\n")
for label, score in zip(result2['labels'], result2['scores']):
    bar = "█" * int(score * 30)
    mark = "✅" if score > 0.5 else "  "
    print(f"  {mark} {label:<20} {score:.4f}  |{bar}|")
````

## 🔬 Section 8 — Named Entity Recognition (NER)

**Model used:** `dbmdz/bert-large-cased-finetuned-conll03-english`  
**Architecture:** Encoder-only (BERT)  
**Task:** Identify and label real-world entities: people, organizations, locations, dates.

````python
print("📥 Loading NER model ...")
ner = pipeline(
    "ner",
    model="dbmdz/bert-large-cased-finetuned-conll03-english",
    aggregation_strategy="simple",
    device=device
)
print("✅ NER model loaded!")
````

````python
text = """
Elon Musk, the CEO of Tesla and SpaceX, announced that the company will expand operations
to Berlin, Germany in early 2025. Apple Inc. and Microsoft also expressed interest in
investing in the European market. The announcement was made at a conference in San Francisco,
California, attended by representatives from the United Nations.
"""

entities = ner(text)

print("🔍 Named Entity Recognition Results:\n")
print(f"Text: {text.strip()}\n")
print(f"{'Entity':<30} {'Type':<15} {'Score'}")
print("─" * 55)

entity_icons = {"PER": "👤", "ORG": "🏢", "LOC": "📍", "MISC": "🔹"}
for ent in entities:
    icon = entity_icons.get(ent['entity_group'], "❓")
    print(f"  {icon} {ent['word']:<28} {ent['entity_group']:<15} {ent['score']:.4f}")
````

## 🖼️ Section 9 — Image Classification

**Model used:** `google/vit-base-patch16-224`  
**Architecture:** Vision Transformer (ViT)  
**Task:** Classify an image into one of 1000 ImageNet categories.

````python
print("📥 Loading image classification model: ViT ...")
img_classifier = pipeline(
    "image-classification",
    model="google/vit-base-patch16-224",
    device=device
)
print("✅ Image classification model loaded!")
````

## 🎨 Section 10 — Image Generation with Stable Diffusion

**Model used:** `CompVis/stable-diffusion-v1-4`  
**Architecture:** Latent Diffusion Model (LDM)  
**Task:** Generate photorealistic images from text descriptions.  

> ⚠️ **Requires GPU** — this will be very slow on CPU. Enable GPU runtime first!

````python
# ── Check if GPU is available ────────────────────────────────────────────────
if not torch.cuda.is_available():
    print("⚠️  No GPU detected. Image generation will be very slow.")
    print("   Go to: Runtime > Change runtime type > T4 GPU")
    print("   Skipping image generation for now.")
    SKIP_IMGGEN = True
else:
    SKIP_IMGGEN = False
    print(f"✅ GPU ready: {torch.cuda.get_device_name(0)}")
    print("   Proceeding with image generation ...")
````

````python
if not SKIP_IMGGEN:
    from diffusers import StableDiffusionPipeline

    print("📥 Loading Stable Diffusion v1.4 (this may take 1–2 minutes)...")
    sd_pipe = StableDiffusionPipeline.from_pretrained(
        "CompVis/stable-diffusion-v1-4",
        torch_dtype=torch.float16,
    )
    sd_pipe = sd_pipe.to("cuda")

    # Memory optimization
    sd_pipe.enable_attention_slicing()
    print("✅ Stable Diffusion loaded!")
````

````python
if not SKIP_IMGGEN:
    prompts = [
        "A futuristic city at sunset with flying cars, highly detailed, digital art",
        "A cute robot reading a book in a cozy library, warm lighting, illustration style",
    ]

    fig, axes = plt.subplots(1, len(prompts), figsize=(10, 5))

    for i, prompt in enumerate(prompts):
        print(f"\n🎨 Generating image {i+1}: '{prompt[:50]}...' ")

        with torch.autocast("cuda"):
            image = sd_pipe(
                prompt,
                num_inference_steps=30,
                guidance_scale=7.5,
                height=512,
                width=512
            ).images[0]

        axes[i].imshow(image)
        axes[i].axis('off')
        axes[i].set_title(f"Prompt {i+1}", fontsize=11)

        # Save image
        image.save(f"/content/generated_image_{i+1}.png")
        print(f"   ✅ Saved to /content/generated_image_{i+1}.png")

    plt.suptitle("🎨 Stable Diffusion — Generated Images", fontsize=14, y=1.02)
    plt.tight_layout()
    plt.show()

    print("\n📌 Prompts used:")
    for i, p in enumerate(prompts, 1):
        print(f"  [{i}] {p}")
````

## 🔤 Section 11 — Fill-Mask (Masked Language Modeling)

**Model used:** `bert-base-uncased`  
**Task:** Predict a missing [MASK] token in a sentence — how BERT was pre-trained.

````python
print("📥 Loading fill-mask model: BERT ...")
fill_mask = pipeline(
    "fill-mask",
    model="bert-base-uncased",
    device=device
)
print("✅ Fill-mask model loaded!")

sentences = [
    "Hugging Face is a [MASK] for machine learning models.",
    "The capital of France is [MASK].",
    "Python is one of the most popular programming [MASK].",
    "The [MASK] architecture revolutionized natural language processing."
]

print("\n🔍 Fill-Mask Predictions:\n")
for sent in sentences:
    results = fill_mask(sent, top_k=3)
    print(f"📌 Input: {sent}")
    print("   Top predictions:")
    for r in results:
        print(f"     → '{r['token_str']}'  (score: {r['score']:.4f})  | {r['sequence']}")
    print()
````

## 🔢 Section 12 — Tokenization Deep Dive

Understanding how text becomes numbers that models can process.

````python
from transformers import AutoTokenizer

tokenizer = AutoTokenizer.from_pretrained("bert-base-uncased")

text = "Transformers make NLP tasks surprisingly easy!"

# Tokenize
tokens = tokenizer.tokenize(text)
encoding = tokenizer(text, return_tensors="pt")
input_ids = encoding['input_ids'][0].tolist()
decoded = [tokenizer.decode([id]) for id in input_ids]

print("🔤 Tokenization Breakdown:\n")
print(f"Original text : {text}")
print(f"\nTokens ({len(tokens)}): {tokens}")
print(f"\nInput IDs    : {input_ids}")
print("\nMapping:")
print(f"  {'Token':<20} {'ID'}")
print("  " + "─" * 30)
for tok, id in zip(decoded, input_ids):
    print(f"  {repr(tok):<20} {id}")

# Decode back
recovered = tokenizer.decode(input_ids, skip_special_tokens=True)
print(f"\nDecoded back  : {recovered}")
````
