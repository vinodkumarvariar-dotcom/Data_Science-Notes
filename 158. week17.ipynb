Loss Function
Write a Python program that:
Computes:
Mean Squared Error (MSE)


Binary Cross-Entropy Loss
Create random examples


```python
import numpy as np

# ----------------------------
# Loss Functions
# ----------------------------

# Mean Squared Error
def mse(y_true, y_pred):
    return np.mean((y_true - y_pred) ** 2)

# Binary Cross Entropy
def binary_cross_entropy(y_true, y_pred):
    eps = 1e-9   # small value to avoid log(0)
    y_pred = np.clip(y_pred, eps, 1 - eps)
    return -np.mean(
        y_true * np.log(y_pred) +
        (1 - y_true) * np.log(1 - y_pred)
    )

# ----------------------------
# Create Random Examples
# ----------------------------

np.random.seed(1)

# True labels (0 or 1)
y_true = np.random.randint(0, 2, size=6)

# Predicted probabilities (between 0 and 1)
y_pred = np.random.rand(6)

print("True Labels:      ", y_true)
print("Predicted Values:", y_pred)

# ----------------------------
# Compute Losses
# ----------------------------

mse_loss = mse(y_true, y_pred)
bce_loss = binary_cross_entropy(y_true, y_pred)

print("\nMean Squared Error:", mse_loss)
print("Binary Cross Entropy Loss:", bce_loss)
```

Build a Simple ANN Using Keras
Using Keras:
Load MNIST dataset


Build the model:
Input → Dense(128, ReLU)
       → Dense(10, Softmax)
optimizer = Adam
loss = SparseCategoricalCrossentropy
Explain Model building steps
Compile → Fit workflow
Role of optimizer and loss


```python
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Flatten
from tensorflow.keras.optimizers import Adam
from tensorflow.keras.losses import SparseCategoricalCrossentropy
```

```python
(x_train, y_train), (x_test, y_test) = tf.keras.datasets.mnist.load_data()

print(x_train.shape)   # (60000, 28, 28)
print(y_train.shape)   # (60000,)
```

```python
x_train = x_train / 255.0
x_test = x_test / 255.0
```

```python
model = Sequential([
    Flatten(input_shape=(28, 28)),     # Input layer
    Dense(128, activation='relu'),     # Hidden layer
    Dense(10, activation='softmax')    # Output layer
])
```

```python
model.compile(
    optimizer=Adam(),
    loss=SparseCategoricalCrossentropy(),
    metrics=['accuracy']
)
```

```python
model.fit(
    x_train, y_train,
    epochs=10,
    batch_size=32,
    validation_split=0.2
)
```

```python
test_loss, test_accuracy = model.evaluate(x_test, y_test)
print("Test Accuracy:", test_accuracy)
```
